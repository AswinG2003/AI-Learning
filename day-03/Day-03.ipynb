```python
import numpy as np

def dot_prod_np(v1, v2):
    return np.dot(v1, v2)

def vector_length_np(v):
    v = np.array(v)
    return np.linalg.norm(v)

def cosine_similarity(v1, v2):
    return dot_prod_np(v1, v2) / (vector_length_np(v1) * vector_length_np(v2))

print(dot_prod_np([1,2,3], [4,5,6]))
print()
print(cosine_similarity([1,2,3], [4,5,6]))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# The function we want to minimise: y = x²
# Minimum is at x=0 where y=0
# Gradient descent should find x=0 automatically

def f(x):
    return x ** 2

def gradient(x):
    return 2 * x  # derivative of x² is 2x

# Gradient descent settings
x = 10.0          # starting point — far from minimum
learning_rate = 0.1
steps = 50

# Store history for plotting
x_history = [x]
y_history = [f(x)]

# Run gradient descent
for i in range(steps):
    grad = gradient(x)      # how steep is the slope here?
    x = x - learning_rate * grad  # take a step downhill
    x_history.append(x)
    y_history.append(f(x))

# Plot the function
x_range = np.linspace(-12, 12, 100)
y_range = f(x_range)

plt.figure(figsize=(12, 5))

# Plot 1 — the path taken
plt.subplot(1, 2, 1)
plt.plot(x_range, y_range, 'b-', label='f(x) = x²')
plt.plot(x_history, y_history, 'ro-', markersize=4, label='gradient descent steps')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Gradient Descent Path')
plt.legend()
plt.grid(True)

# Plot 2 — loss over time
plt.subplot(1, 2, 2)
plt.plot(y_history, 'g-')
plt.xlabel('Step')
plt.ylabel('Loss (f(x))')
plt.title('Loss Decreasing Over Time')
plt.grid(True)

plt.tight_layout()
plt.show()

print(f"Started at x = 10.0")
print(f"Ended at x = {x_history[-1]:.6f}")
print(f"Final loss = {y_history[-1]:.8f}")
```

```python
learning_rates = [0.01, 0.1, 0.9]
plt.figure(figsize=(10, 5))

for lr in learning_rates:
    x = 10.0
    y_hist = []
    for i in range(50):
        x = x - lr * gradient(x)
        y_hist.append(f(x))
    plt.plot(y_hist, label=f'lr={lr}')

plt.xlabel('Steps')
plt.ylabel('Loss')
plt.title('Effect of Learning Rate')
plt.legend()
plt.grid(True)
plt.show()
```

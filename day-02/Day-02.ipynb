```python
# 1. Write a function called circle_stats that takes radius
#    and returns BOTH area and circumference as a tuple
#    area = π × r²
#    circumference = 2 × π × r
#    use 3.14159 for π — no imports needed
```

```python

def circle_stats(r):
    a=3.14159*r**2
    c=2*3.14159*r
    return (a,c)
circle_stats(5)
    
```

```python
area, circumference = circle_stats(5)
print(f"Area: {area}")           # Area: 78.53975
print(f"Circumference: {circumference}")
```

```python
# Step 1 — your 3 functions from Day 1
def dot_product(v1, v2):
    return sum(x * y for x, y in zip(v1, v2))

def vector_length(v):
    return sum(x**2 for x in v) ** 0.5

def cosine_similarity(v1, v2):
    return dot_product(v1, v2) / (vector_length(v1) * vector_length(v2))

# Test
print(cosine_similarity([1,0,0,0], [1,0,0,0]))  # should print 1.0
print(cosine_similarity([1,0,0,0], [0,1,0,0]))  # should print 0.0
```

```python
items = {
    "cat":        [1.0, 0.1, 0.8, 0.2],
    "dog":        [0.9, 0.1, 0.7, 0.3],
    "car":        [0.1, 0.9, 0.0, 0.1],
    "motorcycle": [0.1, 0.8, 0.0, 0.2],
    "tiger":      [0.8, 0.1, 0.6, 0.9],
    "bicycle":    [0.2, 0.7, 0.1, 0.1],
}

```

```python
# Step 1 — define the function (machine blueprint)
def find_similar(query_vector, items_dict, top_n=3):
    score = []
    for key, values in items_dict.items():
        similarity = cosine_similarity(query_vector, values)
        score.append((key, similarity))
    score.sort(key=lambda x: x[1], reverse=True)
    return score[:top_n]

# Step 2 — call the function (turn the machine on)
animal_query = [0.9, 0.1, 0.7, 0.4]
results = find_similar(animal_query, items)
print(results)
vehicle_query = [0.1, 0.85, 0.0, 0.1]
results = find_similar(vehicle_query, items)
print(results)
```

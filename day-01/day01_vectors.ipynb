```python
# Cosine similarity — measures the angle between vectors
# Result: 1.0 = identical, 0 = unrelated, -1 = opposite

def cosine_similarity(v1, v2):
    return dot_product(v1, v2) / (vector_length(v1) * vector_length(v2))

# Represent 3 words as vectors
cat = [1.0, 0.2, 0.8]
dog = [0.9, 0.3, 0.7]
car = [0.1, 0.9, 0.2]

print("cat vs dog:", cosine_similarity(cat, dog))
print("cat vs car:", cosine_similarity(cat, car))
print("dog vs car:", cosine_similarity(dog, car))


```
